Q1. What is a Decision Tree? Explain its structure (Root, Nodes, Leaves) with a real-life example.
Ans:A Decision Tree is a supervised machine learning algorithm used for classification and regression problems. It works like a flowchart where data is split into different branches based on conditions or decisions.

The model learns rules from data and makes predictions by following a path from the top of the tree to the bottom.

Structure of a Decision Tree

A Decision Tree mainly contains three parts:

Root Node The top-most node of the tree. Represents the entire dataset. The first decision or condition starts from here. Example:
“Is the weather sunny?”

Decision Nodes (Internal Nodes) Intermediate nodes where data is divided based on conditions. Each branch represents an outcome of a decision. Example: “Is temperature greater than 30°C?” “Is it raining?”

Leaf Nodes (Terminal Nodes) Final nodes of the tree. Show the final prediction or output. No further splitting happens. Example: “Play Cricket” “Do Not Play Cricket” Real-Life Example: Should We Play Cricket?

Suppose we want to decide whether to play cricket based on weather conditions.

                Weather?
               /        \
           Sunny       Rainy
           /              \
    Humidity High?      Wind Strong?
       /      \            /      \
    Yes       No        Yes       No
     |         |          |         |
 Don't Play   Play     Don't Play  Play
Q2. Differentiate between Gini Impurity and Entropy. Which one is used by default in Scikit-learn and why

Ans: Both Gini Impurity and Entropy are metrics used in a Decision Tree to measure how well a split separates the data into classes.

They help the algorithm choose the best feature for splitting the dataset.

Basis	Gini Impurity	Entropy
Meaning	Measures impurity	Measures randomness/information
Formula Complexity	Simpler	More complex (uses logarithm)
Computation Speed	Faster	Slower
Value Range	0 to 0.5 (binary)	0 to 1 (binary)
Used In	CART algorithm	ID3, C4.5 algorithms
Split Selection	Less computational cost	More accurate in some cases
In Scikit-learn, the default criterion for Decision Trees is:

criterion="gini"
Scikit-learn uses Gini because:

Faster Computation

Gini does not use logarithmic calculations.
Makes training quicker.
Efficiency

Works well for large datasets.
Performance

In most cases, Gini and Entropy produce very similar results.
Gini is computationally less expensive.
Q3. What is Overfitting in Decision Trees? How can we detect it using training and testing accuracy?

Ans:Overfitting in Decision Trees occurs when the model learns the training data too perfectly, including noise and unnecessary details. As a result, it performs very well on training data but poorly on unseen testing data.

We can detect overfitting by comparing training accuracy and testing accuracy:

If both accuracies are similar → the model is performing well.
If training accuracy is very high and testing accuracy is much lower → the model is overfitting.
Q4. Explain Pruning in Decision Trees. What is the difference between Pre-pruning and Post-pruning?

Ans:Pruning in Decision Trees is the process of reducing unnecessary branches of the tree to prevent overfitting and improve model performance on unseen data.

It helps make the tree:

simpler,
faster,
and more accurate for new data.
Difference between pre-pruning and post-pruning are,

Basis	Pre-pruning	Post-pruning
Timing	Stops tree growth early	Removes branches after full growth
Complexity	Simpler and faster	More accurate but slower
Risk	Underfitting possible	Better generalization
Process	Prevents unnecessary splits	Removes unnecessary branches

df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# Categorical columns → fill with mode
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
df[col] = df[col].fillna(df[col].mode()[0])
